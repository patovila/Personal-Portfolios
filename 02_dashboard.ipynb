```python
# Tu portfolio
mis_tickers = ["YPF", "PAM", "TGS", "MELI"]
benchmarks = ["SPY"]  # S&P 500 como referencia
```

```python
# Ver ficha completa de cualquier empresa
mostrar_ticker("YPF")
```

```python
# Tabla de variaciones del portfolio
resultados = [variaciones(t) for t in mis_tickers]
df_var = pd.DataFrame([r for r in resultados if r is not None])
df_var = df_var[["ticker", "precio_actual", "1_dia_%", "1_semana_%", "1_mes_%", "YTD_%", "1_año_%", "5_años_%"]]
df_var
```

```python
# Métricas avanzadas (CAGR, Sharpe, etc.)
metricas = []
for t in mis_tickers:
    m = metricas_avanzadas(t, "5y")
    if m:
        metricas.append({
            "Ticker": m["ticker"],
            "CAGR %": m["CAGR_%"],
            "Vol %": m["volatilidad_anual_%"],
            "Sharpe": m["sharpe_ratio"],
            "Max DD %": m["max_drawdown_%"],
            "% días +": m["%_dias_positivos"],
        })

df_met = pd.DataFrame(metricas).sort_values("Sharpe", ascending=False)
df_met
```
